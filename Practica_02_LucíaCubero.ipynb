## Ejercicio 1.
Crea un diccionario con la siguiente información de tu vida (edad, sexo, estudios, nombre_amigos) [ojo, nombre_amigos será una lista con varios nombres].

```python
diccionario = {
    "edad" : 22,
    "sexo" : "femenino",
    "estudios" : "Grado en Psicología",
    "nombre_amigos" : ["Adri", "Estrella", "Gigi"]
}
```

## Ejercicio 2.
Crea una función que imprima el diccionario anterior.

```python
def imprimir_dic (dic):
  for i in dic:
    value = dic[i]
    print(f"{i}", ":", value)

imprimir_dic(diccionario)
```

## Ejercicio 3.
Crea una clase que reciba como entrada el diccionario y lo imprima.

```python
class Diccionario:
    def __init__(self, dic_comp):
      self.dic_comp = dic_comp
    def imprimir(self):
      for i in self.dic_comp:
        value = self.dic_comp[i]
        print(f"{i}", ":", value)

mi_diccionario = Diccionario(diccionario)

mi_diccionario.imprimir()
```

## Ejercicio 4.
Crea una lista de diccionarios con distintas personas como clase.

```python
class Persona:
  def __init__ (self, nombre, edad, sexo, estudios, nombre_amigos):
    self.nombre = nombre
    self.edad = edad
    self.sexo = sexo
    self.estudios = estudios
    self.amigos = nombre_amigos
  def diccionario (self):
    return {
        "nombre" : self.nombre,
        "edad" : self.edad,
        "sexo" : self.sexo,
        "estudios" : self.estudios,
        "nombre_amigos" : self.amigos
    }

Adri = Persona("Adri", 22, "masculino", "Grado en Psicología", ["Carlota", "Alex", "Minerva"])
Gigi = Persona("Gigi", 22, "femenino", "Grado en Psicología", ["Sofía", "Alex", "Mariló"])
Estrella = Persona("Estrella", 22, "femenino", "Grado en Marketing digital", ["María", "Carlos", "Ivana"])

amigos = [Adri.diccionario(), Gigi.diccionario(), Estrella.diccionario()]
```

## Ejercicio 5.
Imprime la lista anterior con un for().

```python
for persona in amigos:
  nombre = persona["nombre"]
  print(f"Información de {nombre}:", persona)
```

## Ejercicio 10.
Calcula la derivada de x*3-2x y dibuja ambas

```python
x = torch.linspace(-100, 28, 120, requires_grad=True)
y = x*3-2*x
y.backward(torch.ones_like(x))
x_vals = x.detach().numpy()
y_vals = y.detach().numpy()
dy_dx_vals = x.grad.detach().numpy()
plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label="$x*3-2*x$", color="blue")
plt.plot(x_vals, dy_dx_vals, label="Derivada $dy/dx$", color="red", linestyle="--")
plt.legend()
plt.grid(True)
plt.show()
```
